#  Perkalian Titik (Dot Product) Vektor

##  Definisi

Perkalian titik (dot product) adalah operasi aljabar antara dua vektor dengan panjang yang sama, menghasilkan sebuah skalar.

Misal dua vektor:

$$
\mathbf{v} = \begin{bmatrix} v_1 \\ v_2 \end{bmatrix}, \quad
\mathbf{w} = \begin{bmatrix} w_1 \\ w_2 \end{bmatrix}$$

Maka dot product:


$$\mathbf{v} \cdot \mathbf{w} = v_1w_1 + v_2w_2$$

### Contoh

$$
\begin{bmatrix} 2 \\ -3 \end{bmatrix} \cdot \begin{bmatrix} 4 \\ 1 \end{bmatrix} = 2 \cdot 4 + (-3) \cdot 1 = 8 - 3 = 5
$$

---

## Latihan

1. Hitung:

$$
\begin{bmatrix} 3 \\ 4 \end{bmatrix} \cdot \begin{bmatrix} 2 \\ -2 \end{bmatrix}
$$

2. Panjang vektor $\mathbf{v} = \begin{bmatrix} 3 \\ 4 \end{bmatrix}$:

$$
|\mathbf{v}| = \sqrt{3^2 + 4^2} = 5
$$

3. Hitung:

$$
\mathbf{v} \cdot \mathbf{v} = 3 \cdot 3 + 4 \cdot 4 = 25
$$

dan bandingkan dengan $|\mathbf{v}|^2$

4. Rotasi vektor $\mathbf{v}$ sebesar $90^\circ$ CCW:

$$
\text{Matriks rotasi:} \quad
\begin{bmatrix}
0 & -1 \\
1 & 0
\end{bmatrix}
$$

Hasil rotasi:


$$\mathbf{w} = 
\begin{bmatrix}
0 & -1 \\
1 & 0
\end{bmatrix} \begin{bmatrix}
3 \\
4 \end{bmatrix} \begin{bmatrix}
-4 \\
3 \end{bmatrix}
$$


$$\mathbf{v} \cdot \mathbf{w} = 3 \cdot (-4) + 4 \cdot 3 = -12 + 12 = 0
$$

---

## 📐 Geometri Perkalian Titik

Untuk vektor berdimensi $n$:

$$
\mathbf{v} \cdot \mathbf{w} = \sum_{i=1}^{n} v_i w_i
$$

Juga dapat ditulis sebagai:

$$
\mathbf{v} \cdot \mathbf{w} = |\mathbf{v}| |\mathbf{w}| \cos(\theta)
$$

---

## 📎 Sifat-Sifat Dot Product

1. Komutatif:

$$
\mathbf{v} \cdot \mathbf{w} = \mathbf{w} \cdot \mathbf{v}
$$

2. Distributif:

$$
\mathbf{v} \cdot (\mathbf{w} + \mathbf{u}) = \mathbf{v} \cdot \mathbf{w} + \mathbf{v} \cdot \mathbf{u}
$$

3. Skalar:

$$
r(\mathbf{v} \cdot \mathbf{w}) = (r\mathbf{v}) \cdot \mathbf{w}
$$

4. Positif semidefinit:

$$
\mathbf{v} \cdot \mathbf{v} \geq 0
$$

---


## 💻 Program Python


```python
def dot_product(vec1, vec2):
    if len(vec1) != len(vec2):
        raise ValueError("Vektor harus memiliki panjang yang sama.")
    return sum(a * b for a, b in zip(vec1, vec2))

# Contoh penggunaan
v1 = [2, 3, 4]
v2 = [5, 6, 7]
print("Dot Product:", dot_product(v1, v2))
```
